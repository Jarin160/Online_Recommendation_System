```python
import numpy as np

np.random.seed(42)
```

```python
n_arms = 15
n_users = 2500
```

```python
# Hidden truth
true_q = np.random.normal(loc=0, scale=1, size=n_arms)

# Environment response: called EVERY time we recommend
def get_reward(a):
    return np.random.normal(loc=true_q[a], scale=1)
```

```python
print("True qualities:\n", np.round(true_q, 2))
print("True top 5 items:", np.argsort(true_q)[::-1][:5])

# Pull item 0 five times: rewards should differ each time
print("5 rewards from item 0:", [round(get_reward(0), 2) for _ in range(5)])

# Pull item 0 10,000 times: the average should be close to true_q[0]
avg = np.mean([get_reward(0) for _ in range(10000)])
print("true_q[0] =", round(true_q[0], 2), "| average reward =", round(avg, 2))
```

```python
# --- Agent's notebook ---
Q = np.zeros(n_arms)   # Q[a] = my current estimate of item a's quality
N = np.zeros(n_arms)   # N[a] = how many times I've recommended item a

def update(a, r):
    """After recommending item a and seeing reward r, update my notebook."""
    N[a] += 1
    Q[a] += (r - Q[a]) / N[a]
```

```python
epsilon = 0.1

def select_action(Q, epsilon):
    """Pick which item to recommend to the next user."""
    if np.random.rand() < epsilon:
        # EXPLORE: pick any item uniformly at random
        return np.random.randint(n_arms)
    else:
        # EXPLOIT: pick the item with the highest estimate
        # (if several tie, choose randomly among them)
        best_items = np.flatnonzero(Q == Q.max())
        return np.random.choice(best_items)
```

```python
# --- Reset the notebook (clears anything left from earlier tests) ---
Q = np.zeros(n_arms)
N = np.zeros(n_arms)

# --- Logs, so we can analyze and plot later ---
rewards = np.zeros(n_users)               # reward received by each user
actions = np.zeros(n_users, dtype=int)    # item recommended to each user

# --- The main loop: one iteration = one user ---
for t in range(n_users):
    a = select_action(Q, epsilon)   # 1. choose an item
    r = get_reward(a)               # 2. user reacts, we observe a reward
    update(a, r)                    # 3. learn from it
    rewards[t] = r                  # 4. record what happened
    actions[t] = a

# --- Quick look at the results ---
top5 = np.argsort(Q)[::-1][:5]
print("My top 5:  ", top5)
print("True top 5:", np.argsort(true_q)[::-1][:5])
print("Average reward over all users:", round(rewards.mean(), 3))
print("Times each item was recommended:", N.astype(int))
```
